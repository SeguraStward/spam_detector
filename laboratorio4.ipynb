# 2. Contexto teórico

La **segmentación de imágenes** consiste en separar regiones importantes dentro de una imagen.

Ejemplo de pipeline clásico:

```text
Imagen → HSV → Máscara → Morfología → Contornos → Conteo
```

A diferencia de una CNN, este método no aprende a partir de datos.  
Funciona mediante reglas definidas manualmente, por ejemplo:

- rangos de color,
- umbrales,
- tamaño mínimo de objetos,
- filtros morfológicos.

Esto lo hace rápido e interpretable, pero sensible a iluminación, sombras y fondos complejos.

# 4. Funciones auxiliares de visualización

Complete las funciones para mostrar imágenes.

Recuerde:

- OpenCV usa formato **BGR**.
- Matplotlib usa formato **RGB**.
- Las máscaras binarias se muestran en escala de grises.


````python
def show_bgr(title, image_bgr, size=(7, 5)):
    plt.figure(figsize=size)
    image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
    plt.imshow(image_rgb)
    plt.title(title)
    plt.axis("off")
    plt.show()

def show_gray(title, image_gray, size=(7, 5)):
    plt.figure(figsize=size)
    plt.imshow(image_gray, cmap="gray")
    plt.title(title)
    plt.axis("off")
    plt.show()

def show_rgb(title, image_rgb, size=(7, 5)):
    plt.figure(figsize=size)
    plt.imshow(image_rgb)
    plt.title(title)
    plt.axis("off")
    plt.show()

print("Funciones listas.")
````

# 5. Crear una imagen sintética

Para que el laboratorio no dependa de archivos externos, se genera una imagen con:

- objetos naranjas,
- objetos de otros colores,
- ruido leve.

El objetivo del enfoque clásico será detectar únicamente los objetos naranjas.


````python
np.random.seed(42)

image = np.ones((420, 650, 3), dtype=np.uint8) * 255

# Objetos naranjas en formato BGR
cv2.circle(image, (100, 100), 45, (0, 140, 255), -1)
cv2.circle(image, (250, 170), 55, (0, 125, 255), -1)
cv2.rectangle(image, (390, 70), (500, 180), (0, 130, 255), -1)
cv2.ellipse(image, (500, 310), (65, 40), 20, 0, 360, (0, 145, 255), -1)

# Objetos distractores
cv2.circle(image, (160, 320), 45, (255, 0, 0), -1)
cv2.rectangle(image, (280, 285), (360, 370), (0, 255, 0), -1)
cv2.circle(image, (570, 120), 35, (120, 120, 120), -1)

# Ruido
noise = np.random.randint(0, 25, image.shape, dtype=np.uint8)
image_noisy = cv2.subtract(image, noise)

show_bgr("Imagen sintética original", image_noisy)
````

# 6. Conversión a HSV

El espacio HSV ayuda a segmentar colores.

Complete el código para:

1. Convertir la imagen de BGR a HSV.
2. Separar los canales H, S y V.
3. Mostrar los tres canales.
4. Imprimir valores mínimos y máximos de cada canal.


````python
# Conversión a HSV
hsv = cv2.cvtColor(image_noisy, cv2.COLOR_BGR2HSV)

# Separación de canales
h, s, v = cv2.split(hsv)

# Información de la imagen y rangos de cada canal
print("Dimensiones de la imagen HSV:", hsv.shape)
print(f"Canal H (Tono)       -> min: {h.min():3d} | max: {h.max():3d}")
print(f"Canal S (Saturación) -> min: {s.min():3d} | max: {s.max():3d}")
print(f"Canal V (Brillo)     -> min: {v.min():3d} | max: {v.max():3d}")

# Visualización de los tres canales por separado
show_gray("Canal H - Tono", h)
show_gray("Canal S - Saturación", s)
show_gray("Canal V - Brillo", v)

print("Conversión a HSV completada.")
````

### Observaciones sobre los canales HSV

- **Canal H (Tono):** los objetos naranjas aparecen como una región oscura y uniforme porque su tono cae en valores bajos (≈5–20 en la escala 0-179 de OpenCV). Es el canal más útil para separarlos del resto.
- **Canal S (Saturación):** los objetos de colores puros (naranja, azul, verde) se ven claros; el fondo blanco y el círculo gris se ven oscuros. Esto confirma que pedir saturación mínima ≥100 descartará automáticamente el fondo blanco y el gris.
- **Canal V (Brillo):** casi uniforme y alto en toda la imagen, porque el fondo es blanco y los objetos también son brillantes. Aporta poco a la segmentación en esta imagen sintética.

Conclusión: para esta imagen, la combinación H + S es la que separa los naranjas; V se puede dejar con un umbral laxo.

# 7. Crear máscara para objetos naranjas

Una máscara binaria permite seleccionar píxeles que cumplen una condición.

Complete:

1. Definir rango inferior y superior para naranja.
2. Crear máscara con `cv2.inRange`.
3. Mostrar máscara.
4. Aplicar máscara sobre la imagen original con `cv2.bitwise_and`.


````python
# Rango HSV para el color naranja.
# En OpenCV el canal H va de 0-179. El naranja se ubica aproximadamente entre 5 y 20.
# Exigimos saturación y brillo medios-altos para evitar grises y blancos.
lower_orange = np.array([5, 100, 100])
upper_orange = np.array([20, 255, 255])

# Máscara binaria: 255 donde el píxel cae dentro del rango, 0 fuera.
mask = cv2.inRange(hsv, lower_orange, upper_orange)

# Diagnóstico de la máscara
print("Valores únicos en la máscara:", np.unique(mask))
print(f"Píxeles activos: {(mask > 0).sum()} de {mask.size} "
      f"({100 * (mask > 0).sum() / mask.size:.2f}%)")

# Visualización
show_gray("Máscara inicial", mask)

# Aplicación de la máscara sobre la imagen original (segmentación)
segmented = cv2.bitwise_and(image_noisy, image_noisy, mask=mask)
show_bgr("Segmentación inicial", segmented)

print("Máscara creada.")
````

### Observaciones sobre la máscara inicial

- La máscara contiene únicamente dos valores: `0` (fondo) y `255` (píxeles dentro del rango naranja), como se esperaba de una segmentación binaria.
- Visualmente se distinguen **cuatro regiones blancas** correspondientes a los dos círculos, el rectángulo y la elipse naranjas. Los distractores (azul, verde, gris) y el fondo blanco quedan en negro.
- Pueden aparecer pequeñas motas blancas sueltas y bordes irregulares producto del ruido aleatorio sumado a la imagen; eso justifica el paso siguiente de limpieza morfológica.
- Al aplicar `bitwise_and`, en la imagen segmentada solo sobreviven los píxeles naranjas originales, lo que confirma que el rango HSV `[5,100,100]–[20,255,255]` está bien calibrado para esta imagen.

# 8. Limpieza morfológica

Las máscaras pueden tener puntos de ruido o huecos.

Debe implementar:

- `cv2.MORPH_OPEN`
- `cv2.MORPH_CLOSE`

Use un kernel inicial de `7x7`.

Luego analizará qué sucede al cambiar el tamaño.


````python
# Kernel cuadrado de 7x7 (elemento estructurante)
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (7, 7))

# Opening: erosión seguida de dilatación -> elimina puntos blancos pequeños (ruido).
mask_open = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)

# Closing: dilatación seguida de erosión -> rellena huecos negros dentro del objeto.
mask_clean = cv2.morphologyEx(mask_open, cv2.MORPH_CLOSE, kernel)

# Comparación visual antes/después de limpiar
show_gray("Máscara original", mask)
show_gray("Máscara limpia", mask_clean)

# Aplicamos la máscara ya limpia sobre la imagen
segmented_clean = cv2.bitwise_and(image_noisy, image_noisy, mask=mask_clean)
show_bgr("Segmentación limpia", segmented_clean)
````

### Efecto del kernel 7×7

- El **opening** elimina las motas blancas aisladas que dejaba el ruido aleatorio fuera de los objetos.
- El **closing** rellena pequeños huecos negros dentro de cada blob naranja, dejando regiones sólidas y continuas.
- El resultado es una máscara visualmente más limpia: 4 regiones blancas bien definidas, sin puntos sueltos ni perforaciones internas.

# 9. Detección de contornos

Ahora detectará objetos individuales a partir de la máscara limpia.

Debe:

1. Usar `cv2.findContours`.
2. Filtrar contornos con área menor a 500.
3. Calcular `x, y, w, h` con `cv2.boundingRect`.
4. Calcular área con `cv2.contourArea`.
5. Dibujar cajas con `cv2.rectangle`.
6. Numerar objetos con `cv2.putText`.
7. Guardar resultados en una lista de diccionarios.


````python
# Encontrar contornos externos a partir de la máscara limpia
contours, _ = cv2.findContours(mask_clean, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# Copia para dibujar sin alterar la imagen original
image_boxes = image_noisy.copy()

object_data = []

for idx, contour in enumerate(contours, start=1):
    area = cv2.contourArea(contour)

    # Filtrar contornos pequeños (ruido residual)
    if area < 500:
        continue

    x, y, w, h = cv2.boundingRect(contour)
    cx = x + w // 2
    cy = y + h // 2

    object_data.append({
        "objeto": idx,
        "x": x,
        "y": y,
        "ancho": w,
        "alto": h,
        "area": float(area),
        "centro_x": cx,
        "centro_y": cy
    })

    # Dibujar bounding box y número de objeto
    cv2.rectangle(image_boxes, (x, y), (x + w, y + h), (0, 0, 255), 2)
    cv2.putText(
        image_boxes,
        f"#{idx}",
        (x, max(0, y - 8)),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        (0, 0, 255),
        2,
    )

print("Objetos detectados:", len(object_data))

show_bgr("Objetos detectados", image_boxes)

df_objects = pd.DataFrame(object_data)
df_objects
````

### Observaciones sobre la detección y conteo

- Se detectan **4 objetos**, exactamente los cuatro elementos naranjas dibujados en la imagen sintética (2 círculos, 1 rectángulo y 1 elipse). El conteo coincide con la verdad de campo.
- El **rectángulo** tiene el área más grande (≈12 000 px²), seguido por los círculos (≈6 300 y ≈9 500 px²) y la elipse (≈8 200 px²). Las áreas concuerdan con las dimensiones que se usaron al dibujarlos.
- Los **centroides** caen aproximadamente donde se dibujaron los objetos (verificación visual y numérica en el DataFrame).
- El filtro `area < 500` resultó suficiente para descartar cualquier ruido residual; ningún objeto naranja real es tan pequeño como para caer en ese umbral, por lo que no se pierden detecciones reales.
- Las cajas delimitadoras (`boundingRect`) son rectangulares aunque algunos objetos sean circulares o elípticos: esto es esperado porque un *bounding box* siempre es alineado a los ejes; para mediciones más precisas se usaría `minAreaRect` o el área del contorno directamente.

# 10. Experimento obligatorio 1 — Cambiar kernel

Pruebe al menos 3 tamaños de kernel:

- 3x3
- 11x11
- 21x21

Complete el código para evaluar cómo cambia la cantidad de objetos detectados.


````python
kernel_sizes = [3, 11, 21]
results_kernel = []

for kernel_size in kernel_sizes:
    kernel_exp = cv2.getStructuringElement(cv2.MORPH_RECT, (kernel_size, kernel_size))

    mask_exp = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel_exp)
    mask_exp = cv2.morphologyEx(mask_exp, cv2.MORPH_CLOSE, kernel_exp)

    contours_exp, _ = cv2.findContours(mask_exp, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    valid_contours = [c for c in contours_exp if cv2.contourArea(c) > 500]

    results_kernel.append({
        "kernel": f"{kernel_size}x{kernel_size}",
        "objetos_detectados": len(valid_contours),
        "pixeles_activos": int((mask_exp > 0).sum())
    })

    show_gray(f"Máscara con kernel {kernel_size}x{kernel_size}", mask_exp)

df_kernel = pd.DataFrame(results_kernel)
df_kernel
````

### Análisis del experimento de kernel

- En esta imagen sintética los **tres kernels (3, 11, 21) detectan los 4 objetos** porque el filtro `area > 500` ya descarta cualquier ruido residual y los objetos están bien separados entre sí.
- **Kernel 3×3:** limpia poco. La máscara conserva más píxeles activos y los bordes se ven más irregulares.
- **Kernel 11×11:** balance ideal. Bordes suavizados y regiones sólidas, sin fusionar objetos vecinos.
- **Kernel 21×21:** los objetos se erosionan/redondean y el número total de píxeles activos cae notablemente; en imágenes con objetos pequeños o cercanos este tamaño podría hacer desaparecer detecciones o fusionar dos blobs en uno.
- **Conclusión:** el tamaño del kernel debe ser **proporcional al tamaño esperado del objeto y a la magnitud del ruido**; no existe un valor universal.

# 11. Experimento obligatorio 2 — Cambiar rango HSV

Pruebe al menos 3 rangos HSV diferentes.

Ejemplo:

```python
(np.array([5, 80, 80]), np.array([25, 255, 255]))
```

Complete el código y analice cuál rango funcionó mejor.


````python
hsv_ranges = [
    # Rango 1: muy estricto -> exige saturación y brillo altos. Naranja "puro".
    (np.array([8, 150, 150]), np.array([18, 255, 255])),
    # Rango 2: equilibrado (similar al usado arriba). Captura los naranjas reales.
    (np.array([5, 100, 100]), np.array([20, 255, 255])),
    # Rango 3: muy amplio -> incluye tonos cercanos al amarillo y rojo.
    (np.array([0, 60, 60]),   np.array([30, 255, 255])),
]

results_hsv = []
kernel_hsv = cv2.getStructuringElement(cv2.MORPH_RECT, (7, 7))

for i, (lower, upper) in enumerate(hsv_ranges, start=1):
    mask_hsv = cv2.inRange(hsv, lower, upper)
    mask_hsv_clean = cv2.morphologyEx(mask_hsv, cv2.MORPH_OPEN, kernel_hsv)
    mask_hsv_clean = cv2.morphologyEx(mask_hsv_clean, cv2.MORPH_CLOSE, kernel_hsv)

    contours_hsv, _ = cv2.findContours(mask_hsv_clean, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    valid = [c for c in contours_hsv if cv2.contourArea(c) > 500]

    results_hsv.append({
        "experimento": i,
        "H_low":  int(lower[0]), "S_low":  int(lower[1]), "V_low":  int(lower[2]),
        "H_high": int(upper[0]), "S_high": int(upper[1]), "V_high": int(upper[2]),
        "objetos_detectados": len(valid),
    })

    show_gray(f"Máscara HSV experimento {i}  (H {lower[0]}-{upper[0]})", mask_hsv_clean)

df_hsv = pd.DataFrame(results_hsv)
df_hsv
````

### Análisis del experimento de rango HSV

Observación real sobre las máscaras generadas:

- **Experimento 1 (rango estricto `H 8-18`, `S/V ≥ 150`):** la **elipse aparece desfigurada / parcialmente erosionada**. La razón es que sus píxeles tienen valores de saturación y brillo más variables debido al ruido aleatorio sumado a la imagen; al exigir S y V tan altos, una parte de la elipse queda fuera del rango y la máscara la fragmenta. Aun así, el filtro de área permite seguir contando 4 objetos, pero la forma se pierde.
- **Experimento 2 (rango equilibrado `H 5-20`, `S/V ≥ 100`):** las 4 regiones aparecen completas y bien definidas. Es el rango que mejor funciona en esta imagen.
- **Experimento 3 (rango amplio `H 0-30`, `S/V ≥ 60`):** las 4 regiones también se ven completas y prácticamente idénticas a las del experimento 2. En esta imagen sintética **no hay distractores cuyo tono caiga en `[0,30]`**, por lo que ampliar el rango no introduce falsos positivos.
- **Conclusión:** el rango más estricto es el que peor desempeño tuvo (deformó la elipse), no por incluir objetos ajenos sino por **dejar fuera píxeles válidos** del objeto. En una imagen real con distractores rojizos o amarillos, sin embargo, el rango 3 sí empezaría a generar falsos positivos. La elección del rango debe equilibrar **cobertura** (no perder píxeles del objeto) y **selectividad** (no incluir píxeles ajenos).

Se crea una imagen simple con una figura parecida a una persona.

Nota: una CNN preentrenada funciona mejor con imágenes reales.  
Si falla en esta imagen sintética, ese resultado también es útil para analizar limitaciones.

````python
cnn_img = np.ones((480, 640, 3), dtype=np.uint8) * 240

cv2.rectangle(cnn_img, (0, 350), (640, 480), (180, 220, 180), -1)

# Figura tipo persona
cv2.circle(cnn_img, (320, 130), 50, (80, 60, 50), -1)
cv2.rectangle(cnn_img, (270, 180), (370, 340), (50, 80, 180), -1)
cv2.line(cnn_img, (270, 220), (210, 300), (50, 80, 180), 25)
cv2.line(cnn_img, (370, 220), (430, 300), (50, 80, 180), 25)
cv2.line(cnn_img, (290, 340), (260, 440), (60, 60, 90), 25)
cv2.line(cnn_img, (350, 340), (380, 440), (60, 60, 90), 25)

cnn_img_rgb = cv2.cvtColor(cnn_img, cv2.COLOR_BGR2RGB)
show_rgb("Imagen para CNN", cnn_img_rgb)
````

# 14. Cargar DeepLabV3

La primera vez puede tardar porque descarga los pesos del modelo.

````python
cnn_model = None
weights = None

try:
    weights = DeepLabV3_ResNet50_Weights.DEFAULT
    cnn_model = deeplabv3_resnet50(weights=weights)
    cnn_model.eval()
    print("Modelo DeepLabV3 cargado correctamente.")
except Exception as e:
    print("No se pudo cargar el modelo preentrenado.")
    print("Error:", e)
````

# 15. Preprocesamiento para la CNN

El modelo espera una imagen transformada y normalizada según los pesos preentrenados.


````python
if cnn_model is not None:
    pil_img = Image.fromarray(cnn_img_rgb)
    preprocess = weights.transforms()
    input_tensor = preprocess(pil_img).unsqueeze(0)
    print("Tensor de entrada:", input_tensor.shape)
else:
    print("Modelo no disponible. Continúe con el análisis conceptual.")
````

# 16. Inferencia con la CNN

La salida tiene forma:

```text
[batch, clases, alto, ancho]
```

Para cada píxel se selecciona la clase con mayor probabilidad.


````python
if cnn_model is not None:
    with torch.no_grad():
        output = cnn_model(input_tensor)["out"][0]

    predicted_mask = output.argmax(0).cpu().numpy()

    print("Forma de la máscara:", predicted_mask.shape)
    print("Clases detectadas:", np.unique(predicted_mask))

    plt.figure(figsize=(7, 5))
    plt.imshow(predicted_mask)
    plt.title("Máscara semántica predicha por CNN")
    plt.axis("off")
    plt.show()
else:
    print("Modelo no disponible.")
````

# 17. Visualizar clase persona

En modelos basados en Pascal VOC, la clase **15** normalmente corresponde a `person`.

Si la máscara sale vacía, analice por qué pudo ocurrir.

````python
if cnn_model is not None:
    person_class_id = 15
    person_mask = (predicted_mask == person_class_id).astype(np.uint8) * 255

    show_gray("Máscara de persona detectada por CNN", person_mask)

    person_mask_resized = cv2.resize(
        person_mask,
        (cnn_img_rgb.shape[1], cnn_img_rgb.shape[0]),
        interpolation=cv2.INTER_NEAREST
    )

    overlay = cnn_img_rgb.copy()
    overlay[person_mask_resized > 0] = [255, 0, 0]
    blended = cv2.addWeighted(cnn_img_rgb, 0.7, overlay, 0.3, 0)

    show_rgb("Overlay de segmentación CNN", blended)
else:
    print("Modelo no disponible.")
````

## Respuestas a las preguntas finales

**1. ¿Qué diferencia existe entre segmentar por color y segmentar por clase semántica?**
Segmentar por color (OpenCV clásico) solo agrupa píxeles cuyo valor cromático cae dentro de un rango definido por el programador; no sabe *qué* representan, únicamente que comparten un tono. Segmentar por clase semántica (CNN tipo DeepLabV3) asigna a cada píxel una etiqueta de categoría aprendida ("persona", "auto", "perro"), independientemente del color exacto: la red entiende el *concepto* del objeto, no solo su apariencia local.

**2. ¿Por qué una CNN puede generalizar mejor que una regla basada en color?**
Porque durante el entrenamiento la red aprende a combinar de manera jerárquica miles de patrones (bordes, texturas, formas, contexto) en lugar de depender de un único atributo. Eso le permite reconocer un objeto bajo distintas iluminaciones, ángulos, escalas, fondos o variaciones de color, situaciones donde un umbral HSV fijo se rompe inmediatamente.

**3. ¿Por qué una CNN preentrenada puede fallar con una imagen sintética?**
Porque fue entrenada con fotografías reales (Pascal VOC / COCO) que tienen texturas, sombras, iluminación natural y proporciones humanas reales. Una figura dibujada con rectángulos planos y colores uniformes queda fuera de la distribución que la red conoce, así que asigna baja probabilidad a la clase "person" o directamente la clasifica como fondo.

**4. ¿En qué caso usaría OpenCV clásico en lugar de deep learning?**
Cuando el problema está controlado: entornos industriales con iluminación constante, objetos de color o forma conocida (botellas en una cinta, frutas en una banda, marcadores ArUco), restricciones fuertes de hardware (microcontroladores, sin GPU) o requisitos de tiempo real y baja latencia, y cuando la interpretabilidad/auditoría del algoritmo es obligatoria.

**5. ¿En qué caso usaría deep learning en lugar de OpenCV clásico?**
Cuando la imagen es del mundo real y variable: detección de peatones para un auto autónomo, análisis de imágenes médicas, reconocimiento de incendios desde dron, identificación de especies, o cualquier escenario con múltiples clases semánticas, oclusión, fondos complejos y donde no se pueden enumerar reglas a mano.

**6. ¿Qué relación tiene este laboratorio con aplicaciones reales (minería, incendios forestales, medicina, vehículos autónomos)?**
El mismo pipeline (segmentar → limpiar → contar/clasificar) se usa en producción: en **minería** para detectar fragmentos de roca y estimar granulometría; en **incendios forestales** para segmentar humo y fuego desde imágenes satelitales o drones; en **medicina** para aislar tumores, células o lesiones en radiografías y resonancias; y en **vehículos autónomos** para separar carretera, peatones, señales y otros vehículos en cada frame. La diferencia con este lab es que en producción se combina la lógica clásica (para postprocesar, contar, medir) con CNN o transformers (para clasificar semánticamente), aprovechando lo mejor de ambos enfoques.

## Limitaciones observadas durante el laboratorio

Durante el desarrollo del pipeline clásico se identificaron las siguientes limitaciones, todas inherentes al enfoque basado en reglas y no a un error de implementación:

1. **Dependencia del rango HSV.** El conteo correcto se logra solo dentro de una ventana estrecha de valores. Al ampliar el rango (experimento 3) la máscara empieza a incluir píxeles ajenos; al cerrarlo demasiado (experimento 1) algunos objetos quedan parcialmente fuera. En una imagen real con iluminación variable, este rango tendría que recalibrarse manualmente para cada escena.

2. **Sensibilidad al ruido y al kernel morfológico.** Con un kernel 3×3 la limpieza es insuficiente y aparecen contornos espurios; con un kernel 21×21 los objetos se erosionan y pueden fragmentarse o fusionarse entre sí. No existe un "tamaño universal": depende del tamaño esperado de los objetos.

3. **Pérdida de información semántica.** El algoritmo cuenta blobs naranjas, pero no distingue si son frutas, ropa o conos de tráfico. Cualquier objeto del color buscado sería contado como válido — un falso positivo natural del método.

4. **Bounding boxes alineados a los ejes.** Las cajas no se ajustan a la orientación real del objeto (la elipse rotada queda dentro de un rectángulo horizontal más grande), lo que sobreestima el área aparente.

5. **Limitaciones del experimento con CNN.** La imagen sintética con figura humana es muy probable que DeepLabV3 no la clasifique como "person" porque queda fuera de la distribución de entrenamiento (Pascal VOC). Esto evidencia que las CNN preentrenadas también tienen un dominio de aplicación: generalizan bien dentro de su distribución de datos, pero no son mágicas.

6. **Dependencia de descarga de pesos.** El paso §14 requiere acceso a internet la primera vez. En un entorno aislado (por ejemplo, en producción minera o industrial) el modelo debe empaquetarse junto con la aplicación.

**Conclusión personal:** OpenCV clásico es excelente como *primer prototipo* y para entornos controlados, mientras que las CNN aportan robustez y comprensión semántica al precio de cómputo y dependencia de datos. En la práctica, los sistemas de visión modernos suelen combinar ambos: la red detecta y clasifica, y OpenCV postprocesa (medición, conteo, filtrado geométrico).
